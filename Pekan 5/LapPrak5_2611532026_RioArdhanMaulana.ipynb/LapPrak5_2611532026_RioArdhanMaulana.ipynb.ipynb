## 2.1 Dasar Teori
### 1. Python

Python merupakan bahasa pemrograman yang memiliki sintaks yang relatif sederhana dan dapat digunakan untuk membuat berbagai jenis program. Dalam praktikum ini, Python digunakan untuk menerapkan konsep perulangan, percabangan, operator, input, dan pencetakan pola.

### 2. Variabel

Variabel digunakan untuk menyimpan nilai yang dapat digunakan selama program berjalan. Pada program praktikum digunakan beberapa variabel seperti batas_2026, ulang_2026, tinggi_2026.

Nilai dalam variabel dapat digunakan dalam perhitungan, menentukan jumlah perulangan, maupun menentukan posisi karakter dalam pola.

### 3. Input dan Konversi Tipe Data

Fungsi input() digunakan untuk menerima data dari pengguna. Secara default, data yang diterima oleh input() berupa teks. Oleh karena itu, fungsi int() digunakan untuk mengubah input menjadi bilangan bulat agar dapat digunakan dalam operasi perhitungan dan perulangan.

Contohnya digunakan pada program untuk memasukkan nilai batas atau tinggi pola.

### 4. Perulangan for

Perulangan for adalah struktur kontrol dalam Python yang digunakan untuk menjalankan suatu perintah berulang kali berdasarkan elemen dalam suatu urutan atau rentang nilai. Perulangan ini dapat digunakan untuk mencetak angka, mengolah data, dan membentuk pola tertentu.
Contoh:
for i in range(1, 6):
    print(i)
Program tersebut menampilkan angka 1 sampai 5. Perintah print(i) dijalankan sebanyak lima kali karena nilai i dimulai dari 1 hingga 5.

### 5. Fungsi range()

Fungsi range() digunakan untuk menghasilkan urutan bilangan yang dapat digunakan oleh perulangan for.

Bentuk penggunaannya dapat berupa:

range(batas)

atau

range(awal, batas)

serta dapat menggunakan langkah:

range(awal, batas, langkah)

Pada program praktikum terdapat penggunaan range() yang dimulai dari 0 maupun 1. Terdapat juga penggunaan langkah negatif seperti range(n_3003, 0, -1) untuk melakukan perulangan secara menurun.

### 6. Nested for

Nested for merupakan perulangan for yang berada di dalam perulangan for lainnya. Perulangan bersarang dapat digunakan untuk mengolah data berdasarkan baris dan kolom.

Dalam praktikum, nested for digunakan untuk membuat pola bintang, pola titik, deret angka, segitiga, dan pola jam pasir. Perulangan bagian luar biasanya digunakan untuk menentukan baris, sedangkan perulangan bagian dalam digunakan untuk menentukan isi setiap baris.

### 7. Percabangan if, elif, dan else

Percabangan digunakan untuk menjalankan perintah berdasarkan kondisi tertentu. if digunakan untuk memeriksa kondisi pertama, elif digunakan untuk memeriksa kondisi tambahan, sedangkan else digunakan ketika kondisi sebelumnya tidak terpenuhi.

Pada program pola jam pasir, percabangan digunakan untuk menentukan karakter yang dicetak berdasarkan posisi baris dan kolom.

### 8. Operator Aritmatika

Operator aritmatika digunakan untuk melakukan operasi matematika. Operator yang digunakan dalam program praktikum antara lain:

+ untuk penjumlahan.
- untuk pengurangan.
* untuk perkalian.
% untuk mendapatkan sisa pembagian.

Operator tersebut digunakan untuk menghitung jumlah, menentukan posisi karakter, menghitung lebar pola, dan mengatur perubahan nilai dalam perulangan.

### 9. Operator Perbandingan

Operator perbandingan digunakan untuk membandingkan dua nilai dan menghasilkan kondisi benar atau salah. Operator yang digunakan dalam program antara lain:

== untuk memeriksa kesamaan nilai.
!= untuk memeriksa ketidaksamaan nilai.
> untuk memeriksa nilai lebih besar.
< untuk memeriksa nilai lebih kecil.
<= untuk memeriksa nilai lebih kecil atau sama dengan.

Operator perbandingan digunakan dalam percabangan untuk menentukan karakter atau proses yang harus dijalankan.

### 10. Operator Logika or dan and

Operator or digunakan ketika salah satu dari beberapa kondisi sudah cukup untuk memenuhi suatu kondisi. Operator and digunakan ketika beberapa kondisi harus terpenuhi secara bersamaan.

Dalam pola jam pasir, operator or digunakan untuk memeriksa posisi tertentu seperti baris pertama atau baris terakhir, sedangkan and digunakan untuk menentukan posisi karakter di bagian dalam pola.

### 11. Parameter end pada print()

Secara default, print() akan berpindah ke baris berikutnya setelah mencetak suatu nilai. Parameter end digunakan untuk mengubah perilaku tersebut.

Penggunaan end="" membuat hasil berikutnya tetap dicetak pada baris yang sama. Sementara itu, end=" " membuat hasil tetap pada baris yang sama dengan tambahan satu spasi.

Konsep ini digunakan untuk menyusun karakter dan angka agar membentuk pola tertentu.

### 12. Perulangan dan Akumulasi

Perulangan dapat digunakan untuk melakukan proses perhitungan berulang. Salah satunya adalah proses penjumlahan menggunakan variabel yang nilainya terus diperbarui pada setiap perulangan.

Contohnya, suatu variabel dapat dimulai dari nilai 0 kemudian ditambahkan dengan nilai i pada setiap perulangan. Dengan cara tersebut, jumlah akhir dapat diperoleh setelah seluruh perulangan selesai.

### 13. Pencetakan Pola

Pencetakan pola merupakan penerapan nested for untuk menghasilkan susunan karakter atau angka dalam bentuk tertentu. Pola dapat dibentuk dengan mengatur jumlah karakter, jumlah spasi, posisi angka, serta perpindahan baris.

Pada praktikum ini, pola yang dibuat meliputi pola titik dan angka, pola persegi menggunakan karakter *, pola hasil penjumlahan i + j, pola jam pasir, serta pola segitiga.

### 14. Pola Jam Pasir Kristal Palindromik

Pola jam pasir kristal palindromik menggunakan kombinasi perulangan, percabangan, perhitungan, dan pencetakan karakter. Program membentuk bagian atas, poros tengah, dan bagian bawah secara teratur.

Deret angka dibuat secara menurun dan kemudian menaik pada setiap baris. Simbol <*> digunakan sebagai poros kristal, sedangkan karakter #, =, dan | digunakan untuk membentuk bingkai pola.

Pengaturan jumlah spasi pada sisi kiri dan kanan digunakan untuk menjaga keseimbangan serta bentuk pola agar tetap simetris.

```python
# Buat file dengan nama perulangan_for1_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_2026 = int(input("Masukkan jumlah perulangan: "))

for i in range(ulang_2026):
    print("Perulangan ke-", i + 1)
```

### 2.2.3 Program "perulangan_for3_2611532026"

```python
# Buat file dengan nama perulangan_for3_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_2026 = int(input("Masukkan jumlah perulangan: "))

jumlah_2026 = 0
for i_2026 in range(1, ulang_2026 + 1):
    print(i_2026, end=" ")
    jumlah_2026 = jumlah_2026 + i_2026

    if i_2026 < ulang_2026:
        print(" + ", end="")
    else:
        print(" = ", jumlah_2026, end="")
print()
print("Jumlah =", jumlah_2026)
```

**Penjelasan :**
1. Baris 1 berfungsi untuk meminta pengguna memasukkan jumlah perulangan. Fungsi `int()` digunakan untuk mengubah input menjadi bilangan bulat dan menyimpannya dalam variabel `ulang_2026`.
2. Baris 3 berfungsi untuk membuat variabel `jumlah_2026` dengan nilai awal `0`. Variabel ini digunakan untuk menyimpan hasil penjumlahan dari setiap angka yang dihasilkan selama perulangan.
3. Baris 4 berfungsi untuk membuat perulangan menggunakan `for` dengan `range(1, ulang_2026 + 1)`. Perulangan dimulai dari angka 1 dan berakhir pada nilai `ulang_2026`.
4. Baris 5 berfungsi untuk menampilkan nilai `i` pada setiap perulangan. Parameter `end=" "` digunakan agar angka ditampilkan pada baris yang sama dan dipisahkan dengan spasi.
5. Baris 6 berfungsi untuk menjumlahkan nilai `i` ke dalam variabel `jumlah_2026`. Operator `+` digunakan untuk menambahkan nilai perulangan dengan jumlah sebelumnya.
6. Baris 8 berfungsi untuk memeriksa apakah nilai `i` masih lebih kecil dari jumlah perulangan. Kondisi ini digunakan untuk menentukan apakah program masih perlu menampilkan tanda `+`.
7. Baris 9 berfungsi untuk menampilkan tanda `+` apabila masih terdapat angka yang akan dijumlahkan. Parameter `end=""` digunakan agar tanda tersebut tetap berada pada baris yang sama.
8. Baris 10 merupakan bagian `else` yang dijalankan ketika nilai `i` sudah mencapai perulangan terakhir.
9. Baris 11 berfungsi untuk menampilkan tanda `=` dan hasil penjumlahan yang tersimpan dalam variabel `jumlah_2026`.
10. Baris 12 berfungsi untuk membuat baris baru setelah proses perulangan selesai dengan menggunakan `print()` tanpa isi.
11. Baris 13 berfungsi untuk menampilkan hasil akhir penjumlahan dengan format "Jumlah =" diikuti nilai yang tersimpan dalam variabel `jumlah_2026`.

### 2.2.4 Program "nested_for1_2611532026"

```python
# Buat file dengan nama nested_for1_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

batas_2026 = int(input("Masukkan nilai batas:"))
for line in range(1, batas_2026 + 1):
    for j in range(1, (-1 * line + batas_2026) + 1):
        print(".", end=" ")
    print(line)
```

**Penjelasan :**
1. Baris 1 berfungsi untuk meminta pengguna memasukkan nilai batas. Fungsi `int()` digunakan untuk mengubah input menjadi bilangan bulat dan menyimpannya dalam variabel `batas_2026`.
2. Baris 2 berfungsi untuk membuat perulangan luar menggunakan `for`. Variabel `line_2026` dimulai dari 1 sampai `batas_2026`.
3. Baris 3 berfungsi untuk membuat perulangan di dalam perulangan luar menggunakan variabel `j`. Perulangan dimulai dari 1 sampai hasil perhitungan `(-1 * line + batas_2026) + 1`. Perulangan ini digunakan untuk menentukan jumlah titik yang akan dicetak pada setiap baris.
4. Baris 4 berfungsi untuk mencetak karakter `"."` sebanyak jumlah perulangan `j`. Parameter `end=""` digunakan agar setiap titik dicetak pada baris yang sama tanpa berpindah baris.
5. Baris 5 berfungsi untuk mencetak nilai `line_2026` setelah seluruh titik pada baris tersebut selesai dicetak. Perintah `print()` juga membuat program berpindah ke baris berikutnya setelah angka ditampilkan.

### 2.2.5 Program "nested_for2_2611532026"

```python
# Buat file dengan nama nested_for2_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

batas_2026 = int(input("Masukkan nilai batas:"))
for line in range(1, batas_2026 + 1):
    for j in range(1, batas_2026 + 1):
        print("*", end="")
    print()  # pindah ke baris berikutnya
```

3. Baris 3 berfungsi untuk membuat perulangan `for` di dalam perulangan pertama menggunakan variabel `j`. Perulangan ini juga berjalan dari angka 1 sampai `batas_2026`.
4. Baris 4 berfungsi untuk mencetak karakter `"*"` pada setiap perulangan. `end=""` membuat karakter dicetak tanpa berpindah ke baris baru.
5. Baris 5 berfungsi untuk membuat baris baru setelah perulangan bagian dalam selesai, sehingga setiap hasil perulangan `i` ditampilkan pada baris yang berbeda.

### 2.2.6 Program "nested_for3_2611532026"

```python
# Buat file dengan nama nested_for4_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

tinggi_2026 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_2026 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_2026 = tinggi_2026
    c_2026 = a_2026
    lebar_2026 = (2 * tinggi_2026) - 2

    for i in range(1, tinggi_2026 + 1):
        b_2026 = c_2026 + 1

        for j in range(1, lebar_2026 + 1):

            # Baris atas dan bawah
            if i == 1 or i == tinggi_2026:
                if j == 1 or j == lebar_2026:
                    print("#", end="")
                else:
                    print("=", end="")

                    # Baris isi
            else:
                if j == 1 or j == lebar_2026:
                    print("|", end="")
                else:
                    if j == c_2026:
                        print("<", end="")
                    elif j == b_2026:
                        print(">", end="")
                    elif j == (lebar_2026 - c_2026):
                        print("<", end="")
                    elif j == (lebar_2026 - c_2026 + 1):
                        print(">", end="")
                    elif j > b_2026 and j < (lebar_2026 - c_2026):
                        print(".", end="")
                    else:
                        print(" ", end="")

            print()

            #Logika asli java
            a_2026 -= 1
            if a_2026 <= 0:
                c_2026 =(-a_2026) + 2
            else:
                c_2026 = a_2026   
```

### 2.2.8 Program "latihan5_2611532026"

```python
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

tinggi_2026 = int(input("Masukkan tinggi segitiga: "))

for i_2026 in range(1, tinggi_2026 + 1):
    print(" " * (tinggi_2026 - i_2026), end="")

    for j_2026 in range(i_2026):
        print("*", end=" ")
    
    print()
```

**Penjelasan :**
1. Baris 1 berfungsi untuk meminta pengguna memasukkan tinggi segitiga. Fungsi `int()` mengubah input menjadi bilangan bulat dan menyimpannya dalam variabel `tinggi_2026`.

3. Baris 4 berfungsi untuk mencetak spasi sebelum karakter `*`. Jumlah spasi ditentukan oleh `tinggi_2026 - i_2026`, sedangkan `end=""` membuat kursor tetap berada pada baris yang sama.
4. Baris 6 berfungsi untuk membuat perulangan `for` menggunakan variabel `j_2026`. Perulangan berjalan sebanyak nilai `i_2026` untuk menentukan jumlah bintang pada setiap baris.
5. Baris 7 berfungsi untuk mencetak karakter `*` dengan `end=" "`, sehingga setiap bintang tetap berada pada baris yang sama dan dipisahkan oleh spasi.
